```python
import pandas as pd 
from pathlib import Path

raw = Path('../data/raw')
data = {f.stem: pd.read_csv(f) for f in sorted(raw.glob('*.csv'))}
for name , df in data.items():
    print(name, df.shape)
    print(list(df.columns))
    print()
```

```python
for name, df in data.items():
    print('=====', name)
    print('Rows by survey year:')
    print(df['SurveyYear'].value_counts().sort_index().to_string())
    print('Distinct locations:', df['Location'].nunique())
    print('Distinct indicators:', df['Indicator'].nunique())
    print('IsTotal:', df['IsTotal'].value_counts(dropna=False).to_dict())
    print('IsPreferred:', df['IsPreferred'].value_counts(dropna=False).to_dict())
    print('Missing Value %:', round(df['Value'].isna().mean() * 100, 1))
    print('Missing CI %:', round(df['CILow'].isna().mean() * 100, 1))
    print(df['CharacteristicCategory'].value_counts(dropna=False).head(8).to_string())
    print()
```

```python
imm = data['immunization_subnational_ken']
latest = imm['SurveyYear'].max()
locs = sorted(imm.loc[imm['SurveyYear'] == latest, 'Location'].unique())
print(latest, len(locs))
print(locs)
```

```python
provinces = ['Central','Coast','Eastern','North Eastern','Nyanza','Rift Valley','Western']

for name, df in data.items():
    print('=====', name)
    print('Rows by year:', df['SurveyYear'].value_counts().sort_index().to_dict())
    print('Already sorted by SurveyYear:', df['SurveyYear'].is_monotonic_increasing)
    print('Missing CI %:', round(df['CILow'].isna().mean() * 100, 1))
    print('Missing denominator %:', round(df['DenominatorUnweighted'].isna().mean() * 100, 1))
    print('Distinct indicators:', df['Indicator'].nunique())
    if 'iycf' not in name:
        for ind in df['Indicator'].unique():
            print('  -', ind)
    print()
```

```python
imm = data['immunization_subnational_ken']
latest = imm[imm['SurveyYear'] == 2022]
pref = latest[latest['IsPreferred'] == 1]
counties = pref[~pref['Location'].isin(provinces)]

print('2022 rows:', len(latest), '| preferred:', len(pref), '| county-only:', len(counties))
print('Counties:', counties['Location'].nunique())
print('Rows per county-indicator pair:', counties.groupby(['Location','Indicator']).size().value_counts().to_dict())
print(counties['DenominatorUnweighted'].describe())
```

```python
# 1. Is the access file truncated? Rows per indicator in 2014, in indicator order
acc = data['access-to-health-care_subnational_ken']
a14 = acc[acc['SurveyYear'] == 2014]
per_ind = (a14.groupby(['IndicatorOrder', 'Indicator']).size()
              .reset_index(name='rows').sort_values('IndicatorOrder'))
print(per_ind['rows'].describe())
print(per_ind.tail(8).to_string(index=False))

# 2. When do county-level rows begin? Locations per survey year, per file
for name, df in data.items():
    print(name, df.groupby('SurveyYear')['Location'].nunique().to_dict())

# 3. Is the missing denominator just the 'Number of ...' count rows?
for name, df in data.items():
    rate_rows = df[~df['Indicator'].str.startswith('Number of')]
    print(name, 'missing denominator among rate rows %:',
          round(rate_rows['DenominatorUnweighted'].isna().mean() * 100, 1))
    
```

```python
acc  = data['access-to-health-care_subnational_ken']
imm  = data['immunization_subnational_ken']
iycf = data['iycf_subnational_ken']
provinces = ['Central','Coast','Eastern','North Eastern','Nyanza','Rift Valley','Western']

# 1. Truncation test: indicators in earlier access surveys that vanish in 2014
early = set(acc.loc[acc['SurveyYear'] < 2014, 'Indicator'])
in2014 = set(acc.loc[acc['SurveyYear'] == 2014, 'Indicator'])
gone = sorted(early - in2014)
print(len(gone), 'indicators in earlier surveys but absent in 2014:')
for i in gone: print('  -', i)
print(acc[['SurveyYear','Location','Indicator']].tail(3).to_string())

# 2. County filter on immunization 2022
latest = imm[imm['SurveyYear'] == 2022]
pref = latest[latest['IsPreferred'] == 1]
counties = pref[~pref['Location'].isin(provinces)]
print('2022 rows:', len(latest), '| preferred:', len(pref), '| county-only:', len(counties))
print('Counties:', counties['Location'].nunique())
print('Rows per county-indicator pair:', counties.groupby(['Location','Indicator']).size().value_counts().to_dict())
print(counties['DenominatorUnweighted'].describe())

# 3. IYCF: which indicators have the missing sample sizes?
print(iycf.groupby('Indicator')['DenominatorUnweighted']
          .apply(lambda s: round(s.isna().mean() * 100, 1)).to_string())
```
